```python
# 2. Create the Catalog dynamically (Fulfills Task 2)
# Note: If your workspace has strict permissions preventing catalog creation, 
# this line will fail, signaling you to change the catalog name in '00_config' to a default catalog.
spark.sql(f"CREATE CATALOG IF NOT EXISTS `{catalog_name}`;")
print(f"📦 Catalog verification complete: `{catalog_name}` is ready.")

# 3. Create the Four Schemas with One-Sentence Comments (Fulfills Tasks 3 & 4)
spark.sql(f"""
  CREATE SCHEMA IF NOT EXISTS `{catalog_name}`.`{schema_bronze}`
  COMMENT 'Stores raw, unaltered source data exactly as it arrives from files.';
""")

spark.sql(f"""
  CREATE SCHEMA IF NOT EXISTS `{catalog_name}`.`{schema_silver}`
  COMMENT 'Houses cleansed, standardized, and conformed production-ready datasets.';
""")

spark.sql(f"""
  CREATE SCHEMA IF NOT EXISTS `{catalog_name}`.`{schema_gold}`
  COMMENT 'Contains business-level aggregations and star schemas optimized for BI tools.';
""")

spark.sql(f"""
  CREATE SCHEMA IF NOT EXISTS `{catalog_name}`.`{schema_ops}`
  COMMENT 'Tracks technical pipeline metadata, operational logging, and execution audit trails.';
""")

print("📁 All four schemas (bronze, silver, gold, ops) created successfully.")

# 4. Verification Check: Show all schemas in the configured catalog (Fulfills Task 5)
print("\n🔍 Verification: Fetching available schemas...")
schemas_df = spark.sql(f"SHOW SCHEMAS IN `{catalog_name}`;")
display(schemas_df)
```
